```python
account_type=input("please enter your account type basic or premium: "). lower()
current_balance=float(input("please enter your current balance: "))
withdraw_amount=float(input("please enter the amount you want to withdraw: "))
if account_type=="premium" and withdraw_amount==15000 or account_type=="basic" and withdraw_amount==5000:
    print("your withdrawal is successful")
    if current_balance>=withdraw_amount:
        print("withdraw success") 
        if withdraw_amount %100 == 0:
            current_balance=current_balance-withdraw_amount
            print("Your new balance is: ",current_balance)
            print("Withdrawal successful")
        else:
            print("Withdrawal amount must be in multiples of 100")
            
    else:
        print("Insufficient balance")

else:
    print("Invalid account type or withdrawal amount")
```
